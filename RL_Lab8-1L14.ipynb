```python
# Assignment 8: State Value and Action Value
# Reinforcement Learning

GAMMA = 0.9

# States
S0 = (0, 0)
S1 = (0, 1)
G  = (0, 2)
S3 = (1, 1)
S4 = (1, 2)

OBSTACLE = (1, 0)

# Policy given in the assignment
policy = {
    S0: "Right",
    S1: "Down",
    S3: "Right",
    S4: "Up"
}

# Move function
def move(state, action):
    r, c = state

    if action == "Up":
        next_state = (r - 1, c)
    elif action == "Down":
        next_state = (r + 1, c)
    elif action == "Left":
        next_state = (r, c - 1)
    elif action == "Right":
        next_state = (r, c + 1)

    # Grid boundary / obstacle
    if (next_state[0] < 0 or next_state[0] > 1 or
        next_state[1] < 0 or next_state[1] > 2 or
        next_state == OBSTACLE):
        return state, -1

    # Goal state
    if next_state == G:
        return next_state, 10

    # Normal movement
    return next_state, -1


# Calculate discounted return
def calculate_return(start_state, first_action=None):
    state = start_state
    total_return = 0
    discount = 1.0

    # First action can be specified for Q(s,a)
    if first_action is not None:
        next_state, reward = move(state, first_action)

        total_return += discount * reward
        discount *= GAMMA
        state = next_state

    # Follow policy
    while state != G:
        action = policy[state]

        next_state, reward = move(state, action)

        total_return += discount * reward
        discount *= GAMMA
        state = next_state

    return total_return


# State Value V(S0)
V_S0 = calculate_return(S0)

# State Values of all states
V_S1 = calculate_return(S1)
V_S3 = calculate_return(S3)
V_S4 = calculate_return(S4)

# Action Value Q(S0, Up)
Q_S0_Up = calculate_return(S0, "Up")


print("State Values:")
print("V(S0) =", round(V_S0, 4))
print("V(S1) =", round(V_S1, 4))
print("V(S3) =", round(V_S3, 4))
print("V(S4) =", round(V_S4, 4))

print("\nAction Value:")
print("Q(S0, Up) =", round(Q_S0_Up, 4))
```
